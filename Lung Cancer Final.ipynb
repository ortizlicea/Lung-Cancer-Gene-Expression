```python
df['-log10_p'] = -np.log10(df['Cluster 1 P-Value']) # creating a column for the -log10 transformation

sns.scatterplot(data=df,  # plotting the points using a scatterplot
                x='Cluster 1 Log2 Fold Change', 
                y='-log10_p'
)

top_genes = df.nsmallest(5, 'Cluster 1 P-Value') # this code is finding the top genes through the 5 with the smallest p-value (highest significance)
for i, row in top_genes.iterrows():
    plt.text(row['Cluster 1 Log2 Fold Change'], row['-log10_p'], row['FeatureName'], fontsize=9)

plt.xlim(-5, 5) # this simply sets the x-axis limits for the plot, since y varies a lot I did not set a limit for it


plt.axhline(y=-np.log10(0.05), color='gray', linestyle='--', alpha=0.5) # this does our horizontal threshold


plt.axvline(x=1, color='gray', linestyle='--', alpha=0.5) # this sets our vertical thresholds at x=1, x=-1
plt.axvline(x=-1, color='gray', linestyle='--', alpha=0.5)

top_genes.FeatureName # lastly this lists out the feature names at the top, a few would overlap and become hard to read, so this made it easier to identify them
```

### Finally, I settled on the following code once I discovered I could simply combine the columns of data into one rather than plotting multiple.

#### I began with the Tumor clusters

```python
# The melt command is one I found online when searching for how to combine columns of a DataFrame. melt does just that! Here I am 
# combining the columns of the Log2 Fold Change of all the clusters in the tumor category.
lfc_df = df.melt(id_vars=['FeatureName'], 
                 value_vars=['Cluster 4 Log2 Fold Change', 'Cluster 6 Log2 Fold Change', 'Cluster 10 Log2 Fold Change', 'Cluster 11 Log2 Fold Change', 'Cluster 12 Log2 Fold Change', 'Cluster 14 Log2 Fold Change', 'Cluster 15 Log2 Fold Change', 'Cluster 16 Log2 Fold Change', 'Cluster 22 Log2 Fold Change'],
                 var_name='Cluster', 
                 value_name='Log2FC')

# Here same thing but "melting" the p-value columns
p_df = df.melt(id_vars=['FeatureName'], 
               value_vars=['Cluster 4 P-Value', 'Cluster 6 P-Value', 'Cluster 10 P-Value', 'Cluster 11 P-Value', 'Cluster 12 P-Value', 'Cluster 14 P-Value', 'Cluster 15 P-Value', 'Cluster 16 P-Value', 'Cluster 22 P-Value'],
               var_name='Cluster_P', 
               value_name='PValue')

# This is just code for a -log10p transformation since p-values are too small for the visualization
lfc_df['-log10_p'] = -np.log10(p_df['PValue'])
```

```python
# now just plot
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=lfc_df,
    x='Log2FC',
    y='-log10_p',
    alpha=0.5
)

# Adding horizontal threshold lines
plt.axhline(y=-np.log10(0.05), color='gray', linestyle='--', alpha=0.5)

# Adding vertical threshold lines

# Add your vertical threshold lines
plt.axvline(x=1, color='gray', linestyle='--')
plt.axvline(x=-1, color='gray', linestyle='--')

plt.title('Combined Volcano Plot: Tumor')
plt.show()
```
